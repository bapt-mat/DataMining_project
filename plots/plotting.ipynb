```python
import matplotlib.pyplot as plt
from PIL import Image

def plot_spectro(file):
    image_path = file
    image = Image.open(image_path)

    width, height = image.size

    plt.imshow(image)
    plt.xlabel('Time (seconds)')
    plt.ylabel('Frequency (Hz)')

    x_ticks = [0, width]
    x_labels = ['0', '1.4']

    y_ticks = [0, height]
    y_labels = ['6000', '500']

    plt.xticks(ticks=x_ticks, labels=x_labels)
    plt.yticks(ticks=y_ticks, labels=y_labels)

    plt.axis('on')
    plt.show()
```

```python
file = 'sample_voc_spectro_12_5.png'
plot_spectro(file)
```

```python
file = 'sample_noise_spectro_12_5.png'
plot_spectro(file)
```
